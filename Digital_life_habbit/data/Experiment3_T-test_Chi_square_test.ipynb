Hypothesis testing is a statistical method used to evaluate a claim about a population using evidence from sample data.

```python
df=pd.read_csv(r'c:\\Users\\HP\\Downloads\\Dataset_digital_life_impact_on_habits\\social_media_sleep_stress_productivity_11000.csv')
```

Is average productivity different between Low-stress and High-stress people

```python
low = []
high = []

for i in range(len(df)):

    stress = df.iloc[i]["stress_level"]
    score = df.iloc[i]["productivity_score"]

    if pd.notna(score):

        if stress == "Low":
            low.append(float(score))

        elif stress == "High":
            high.append(float(score))
```

```python
n1 = len(low)
n2 = len(high)

mean1 = sum(low) / n1
mean2 = sum(high) / n2

print("Low stress mean:", mean1)
print("High stress mean:", mean2)
```

```python
sum1 = 0
sum2 = 0

for value in low:
    sum1 += (value - mean1) ** 2

for value in high:
    sum2 += (value - mean2) ** 2

var1 = sum1 / (n1 - 1)
var2 = sum2 / (n2 - 1)

print("Variance 1:", var1)
print("Variance 2:", var2)
```

```python
import math

standard_error = math.sqrt(
    var1 / n1 + var2 / n2
)

t_stat = (mean1 - mean2) / standard_error

print("t-statistic:", t_stat)
```

Compaare with alpha=0.05,its value=1.96 , t-value is less than 1.96

### Interpretation of t-Test Result

An independent-samples  t-test was performed to compare mean productivity scores between Low-stress and High-stress observations.

The calculated t-statistic was approximately 0.4471.

At the 5% significance level, this value is smaller in absolute magnitude than the approximate two-tailed critical value of 1.96 for large degrees of freedom.

Therefore, we fail to reject the null hypothesis.

**Conclusion:** The analysis does not provide sufficient statistical evidence of a difference in mean productivity between the Low-stress and High-stress groups.

CHI SQUARE TEST

Is a person's social-media platform associated with their stress level?


H0:{Platform and stress level are independent}
H1:{Platform and stress level are associated}


```python
platforms = ["Instagram", "YouTube", "Snapchat",
             "Twitter", "LinkedIn"]

stress_levels = ["Low", "Medium", "High"]
```

```python
observed = []

for platform in platforms:
    row = []

    for stress in stress_levels:
        count = 0

        for i in range(len(df)):
            if (df.iloc[i]["platform"] == platform and
                df.iloc[i]["stress_level"] == stress):
                count += 1

        row.append(count)

    observed.append(row)

print("Observed frequencies:")

for i in range(len(platforms)):
    print(platforms[i], observed[i])
```

```python
row_totals = []

for row in observed:
    row_totals.append(sum(row))

column_totals = []

for j in range(len(stress_levels)):
    total = 0

    for i in range(len(platforms)):
        total += observed[i][j]

    column_totals.append(total)

grand_total = sum(row_totals)

print("Row totals:", row_totals)
print("Column totals:", column_totals)
print("Grand total:", grand_total)
```

```python
expected = []

for i in range(len(platforms)):
    row = []

    for j in range(len(stress_levels)):

        E = (
            row_totals[i] * column_totals[j]
        ) / grand_total

        row.append(E)

    expected.append(row)

print("Expected frequencies:")

for i in range(len(platforms)):
    print(platforms[i], expected[i])
```

```python
chi_square = 0

for i in range(len(platforms)):
    for j in range(len(stress_levels)):

        O = observed[i][j]
        E = expected[i][j]

        chi_square += ((O - E) ** 2) / E

print("Chi-square statistic:", chi_square)
```

```python
r = len(platforms)
c = len(stress_levels)

df_chi = (r - 1) * (c - 1)

print("Degrees of freedom:", df_chi)
```

```python
critical_value = 15.507

if chi_square > critical_value:
    print("Reject H0")
    print("Platform and stress level are associated.")
else:
    print("Fail to reject H0")
    print("Insufficient evidence of association.")
```

platform–stress frequency differences were not large enough, relative to the expected frequencies, to reject independence at the 5% significance level.
