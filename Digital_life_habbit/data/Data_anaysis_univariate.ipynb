## Univariate Exploratory Data Analysis

### Numerical Variable Analysis

Univariate analysis examines one variable at a time to understand
its distribution, concentration, variability and potential outliers.

Histograms are used to examine the frequency distribution of
numerical variables.

## Univariate Visual Analysis

### 1. Distribution of Sleep Duration

**Research Question:** How are daily sleep durations distributed
across the dataset?

**Method:** Histogram

**Objective:** Examine the distribution shape, concentration,
spread, and possible unusual patterns.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\HP\Downloads\Dataset_digital_life_impact_on_habits\social_media_sleep_stress_productivity_11000.csv")
```

```python
plt.hist(df["sleep_hours"], bins=12,edgecolor='black')
plt.title("Distribution of sleep hours")
plt.xlabel("Sleep hours")
plt.ylabel("Frequency")
plt.show()
```

uniform distribution

connect this to statistical understanding
 mean sleep duration is 5.99 hours.

But does that mean most users sleep approximately 6 hours?

No! histogram suggests that sleep durations are distributed relatively evenly between 3 and 9 hours.

This is an excellent example of why we need both descriptive statistics and visualizations.

```python
plt.hist(df["daily_screen_time_hours"], bins=11,edgecolor="black")

plt.title("Distribution of Daily Screen Time")
plt.xlabel("Daily Screen Time (Hours)")
plt.ylabel("Number of Observations")

plt.show()
```

For a continuous uniform distribution between 1 and 12, the theoretical standard deviation is:

sigma = (b-a)/root(n) ~ 3.18hours   
#for n =12


### Interpretation — Daily Screen Time Distribution

The histogram shows an approximately uniform distribution of daily screen time, ranging from 1 to 12 hours.

The frequencies across the intervals are relatively similar, with no strong concentration around a particular screen-time duration.

The mean and median are **6.56 and 6.57 hours**, respectively. The standard deviation is **3.18 hours**, indicating considerable variation in daily screen usage.

The middle 50% of observations lie between **3.81 and 9.33 hours** (IQR = 5.52 hours).

**Key Finding:** Daily screen time is distributed relatively evenly across the observed range. The average of 6.56 hours does not imply that most users spend approximately that amount of time on screens.

This is a descriptive finding and does not establish a relationship between screen time, stress or productivity.

```python
plt.hist(df["social_media_hours"], bins=15,edgecolor="black")

plt.title("Distribution of Daily Social Media Usage")
plt.xlabel("Social Media Usage (Hours)")
plt.ylabel("Number of Observations")

plt.show()
```

### Interpretation — Daily Social Media Usage

The histogram shows an approximately uniform distribution of daily social-media usage, ranging from 0.5 to 8 hours.

The frequencies across the intervals are relatively similar, with no strong concentration around a particular usage duration. A slightly higher frequency is visible around 5–5.5 hours, but the overall distribution remains relatively flat.

The mean and median are both **4.25 hours**, while the standard deviation is **2.16 hours**.

The middle 50% of observations fall between **2.38 and 6.10 hours** (IQR = 3.72 hours).

**Key Finding:** Social-media usage varies across observations and is distributed relatively evenly throughout the recorded range. The mean of 4.25 hours does not imply that most users spend exactly that amount of time on social media.

The potential logical inconsistencies identified between social-media usage and daily screen time should be considered in subsequent relationship analysis.
