```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\HP\Downloads\Dataset_digital_life_impact_on_habits\social_media_sleep_stress_productivity_11000.csv")
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    df["sleep_hours"],
    df["productivity_score"],
    alpha=0.3
)

plt.xlabel("Sleep Duration (Hours)")
plt.ylabel("Productivity Score")
plt.title("Sleep Duration vs Productivity")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["social_media_hours"],
    df["productivity_score"],
    alpha=0.2
)

plt.xlabel("Social Media Usage (Hours)")
plt.ylabel("Productivity Score")
plt.title("Social Media Usage vs Productivity")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["exercise_minutes"],
    df["productivity_score"],
    alpha=0.2
)

plt.xlabel("Exercise Duration (Minutes)")
plt.ylabel("Productivity Score")
plt.title("Exercise Duration vs Productivity")

plt.show()
```
