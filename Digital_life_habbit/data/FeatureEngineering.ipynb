```python
import pandas as pd
df=pd.read_csv(r'c:\\Users\\HP\\Downloads\\Dataset_digital_life_impact_on_habits\\social_media_sleep_stress_productivity_11000.csv')

# Feature 1: Non-social screen time
df["non_social_screen_hours"] = (
    df["daily_screen_time_hours"] -
    df["social_media_hours"]
)

# Feature 2: Sleep categories
df["sleep_category"] = pd.cut(
    df["sleep_hours"],
    bins=[0, 5, 7, 24],
    labels=["Short", "Moderate", "Long"],
    include_lowest=True
)

# Feature 3: Screen-time categories
df["screen_time_category"] = pd.cut(
    df["daily_screen_time_hours"],
    bins=[0, 4, 8, 24],
    labels=["Low", "Moderate", "High"],
    include_lowest=True
)

display(df.head())
```

```python
import numpy as np

# Identify inconsistent observations
condition = (
    df["social_media_hours"] >
    df["daily_screen_time_hours"]
)

print("Before cleaning:", condition.sum())

# Median from unflagged observations
median_social = df.loc[
    ~condition, "social_media_hours"
].median()

# Replace flagged values using constrained median
df.loc[condition, "social_media_hours"] = np.minimum(
    median_social,
    df.loc[condition, "daily_screen_time_hours"]
)

# Recreate the engineered feature
df["non_social_screen_hours"] = (
    df["daily_screen_time_hours"] -
    df["social_media_hours"]
)

print("After cleaning:",
      (df["social_media_hours"] >
       df["daily_screen_time_hours"]).sum())

print("Negative non-social screen values:",
      (df["non_social_screen_hours"] < 0).sum())
```

```python
print("Negative non-social screen values:",
      (df["non_social_screen_hours"] < 0).sum())

print("\nSleep categories:")
print(df["sleep_category"].value_counts())

print("\nScreen-time categories:")
print(df["screen_time_category"].value_counts())
```
