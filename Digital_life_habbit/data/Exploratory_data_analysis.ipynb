```python
numerical_columns = [
    "age",
    "daily_screen_time_hours",
    "social_media_hours",
    "sleep_hours",
    "exercise_minutes",
    "study_work_hours",
    "productivity_score"
]
```

```python
mean=df[numerical_columns].mean()
print(mean)
```

```python
for columns in numerical_columns:
    values=df[columns].dropna().tolist()

    total=0
    for value in values:
        total+=value
    mean_columns=total/len(values)
    for columns in numerical_columns:
        values = df[columns].dropna().tolist()

        total = 0
        for value in values:
            total += value
        mean_columns = total / len(values)
        print(f"Mean of {columns}: {mean_columns:.2f}")
```
