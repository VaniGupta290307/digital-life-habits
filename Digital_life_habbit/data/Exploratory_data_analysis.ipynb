```python
df=pd.read_csv(r"C:\Users\HP\Downloads\Dataset_digital_life_impact_on_habits\social_media_sleep_stress_productivity_11000.csv")
```

```python
numerical_columns = [
    "age",
    "daily_screen_time_hours",
    "social_media_hours",
    "sleep_hours",
    "exercise_minutes",
    "study_work_hours",
    "productivity_score"
]
```

```python
mean=df[numerical_columns].mean()
print(mean)
```

```python
for columns in numerical_columns:
    values=df[columns].dropna().tolist()

    total=0
    for value in values:
        total+=value
    mean_columns=total/len(values)
    print(f"Mean of columns {columns} : {round(mean_columns,2)}")
```

```python
for columns in numerical_columns:
    values=df[columns].dropna().tolist()
    sorted_values=sorted(values)
    n=len(sorted_values)
    if n==0:
        median_columns=None
    elif n%2==1:
        median_columns=sorted_values[(n)//2]
    else:
        median_columns=(sorted_values[(n//2)-1]+sorted_values[n//2])/2
    print(f"Median of columns {columns} : {median_columns}")
```

```python
for columns in numerical_columns:
    column = list(df[columns].dropna())

    frequency = {}

    for value in column:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1
    # for value, count in frequency.items():
        #print(f"Value: {value}, Frequency: {count}")
    max_frequency = max(frequency.values())
    print(f"Max frequency is {max_frequency} for column {columns}")


    # Find the highest frequency
    # Find all values having that frequency
    # Print the mode
```

```python
for column in numerical_columns:
    values = list(df[column].dropna())
    n = len(values)

    if n < 2:
        print(f"Variance of {column}: Insufficient observations")
        continue

    # Calculate mean using formula
    total = 0
    for value in values:
        total += value

    mean = total / n

    # Calculate sum of squared deviations
    squared_deviation_sum = 0

    for value in values:
        squared_deviation_sum += (value - mean) ** 2

    # Sample variance
    variance = squared_deviation_sum / (n - 1)

    print(f"Variance of {column}: {round(variance, 2)}")


    std_dev = variance ** 0.5
    print(f"Standard Deviation of {column}: {round(std_dev, 2)}")

display(df[numerical_columns].var())
display(df[numerical_columns].std())
```
