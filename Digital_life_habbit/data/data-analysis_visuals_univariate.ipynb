```python
stress_counts = df["stress_level"].value_counts()

print(stress_counts)
```

```python
import matplotlib.pyplot as plt
stress = list(df["stress_level"])

frequency = {}

for value in stress:
    if value in frequency:
        frequency[value] += 1
    else:
        frequency[value] = 1

print(frequency)

plt.bar(frequency.keys(), frequency.values())

plt.xlabel("Stress Level")
plt.ylabel("Number of Observations")
plt.title("Distribution of Stress Levels")

plt.show()
```

```python
total = len(stress)

for category, count in frequency.items():
    percentage = (count / total) * 100

    print(f"{category}: {count} ({percentage:.2f}%)")
```

## Box Plot Analysis — Distribution and Potential Outliers

### Objective
To visualize the median, quartiles, interquartile range,
and potential outliers in numerical variables.

A box plot (or box-and-whisker plot) is a statistical visualization that summarizes the distribution of numerical data using the median, quartiles, interquartile range, whiskers and potential outliers.

In simple language:
A box plot helps us understand where the middle 50% of observations lie, how spread out the data is, and whether unusually extreme values are present.

```python
print(df["sleep_hours"].agg(["max","min"]))
plt.boxplot(df["sleep_hours"].dropna())

plt.title("Box Plot of Sleep Duration")
plt.ylabel("Sleep Duration (Hours)")

plt.show()
```

### Interpretation — Sleep Duration Box Plot

The box plot summarizes the distribution of daily sleep duration using quartiles, the median, whiskers and potential outliers.

The median sleep duration is **6 hours**, while Q1 and Q3 are **4.50 and 7.49 hours**, respectively. Therefore, the middle 50% of observations span approximately **2.99 hours**.

The median is positioned near the center of the box, and the whiskers are approximately equal in length, suggesting a relatively balanced distribution.

The whiskers extend from **3 to 9 hours**, and no observations are flagged as outliers under the default 1.5 × IQR rule.

**Key Finding:** Sleep durations show a relatively balanced spread across the observed range, with no potential outliers identified by the box plot method.

This complements the approximately uniform distribution observed in the sleep-duration histogram.

```python
print(df["productivity_score"].agg(["max","min"]))
plt.boxplot(df["productivity_score"].dropna())
plt.title("Box Plot of Productivity Scores")
plt.ylabel("Productivity Score")
plt.show()
```

### Interpretation — Productivity Score Box Plot

The box plot summarizes the distribution of productivity scores using the median, quartiles, interquartile range and whiskers.

The median productivity score is approximately **50.22**, while the first and third quartiles are **25.73 and 76.23**, respectively. The middle 50% of observations therefore span **50.50 points**.

The median is positioned near the center of the box, and the whiskers have relatively similar lengths, suggesting a fairly balanced distribution.

The lower and upper outlier fences are approximately **−50.02 and 151.98**, respectively. Since the recorded scores range from approximately 1 to 99.99, no potential outliers are identified under the 1.5 × IQR rule.

**Key Finding:** Productivity scores show substantial variation across the dataset, but no observations are flagged as statistical outliers using this method.

This finding is consistent with the approximately uniform distribution observed in the productivity histogram.

Further analysis is needed to investigate whether productivity scores are associated with digital usage, sleep, exercise or stress.
